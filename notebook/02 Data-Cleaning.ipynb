```python
import ast

def convert_skills(x):
    if isinstance(x, list):
        return ' '.join(x)

    if isinstance(x, str):
        try:
            parsed = ast.literal_eval(x)
            if isinstance(parsed, list):
                return ' '.join(parsed)
        except (ValueError, SyntaxError):
            return x

    return ''

new_df['job_description'] = (
    new_df['description_text'].fillna('')
    + ' '
    + new_df['ai_key_skills'].apply(convert_skills) +' '+ new_df['ai_requirements_summary']
)
```

```python
def extract_skills(text):
    if not isinstance(text, str):
        return []

    doc = nlp(text)

    return list(dict.fromkeys(
        ent.text
        for ent in doc.ents
        if "SKILLS" in ent.label_.upper()
    ))
```
